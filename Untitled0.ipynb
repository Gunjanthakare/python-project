```python
import sys
!{sys.executable} -m pip install faiss-cpu
!{sys.executable} -m pip install -U bitsandbytes>=0.46.1 accelerate

import json
import re
import torch
import faiss
import numpy as np
import gradio as gr

from datasets import load_dataset
from sentence_transformers import SentenceTransformer
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    pipeline,
    BitsAndBytesConfig
)

print("🔄 Loading data and model... Please wait...")

# ============================================================
# 1. LOAD BNS + BNSS + BSA
# ============================================================

ds = load_dataset(
    "GSMS-B/indian-legal-sections-bns-bnss-bsa-2023",
    split="train"
)

documents = []
metadata = []

for item in ds:
    act = str(item.get("act", "BNS")).strip()
    section = str(item.get("section", item.get("Section", ""))).strip()
    title = str(item.get("title", item.get("Title", ""))).strip()
    content = str(
        item.get("content", item.get("text", item.get("Content", "")))
    )[:700]

    text = f"{act} Section {section} - {title}: {content}"
    documents.append(text)
    metadata.append({
        "act": act,
        "section": section if section else "N/A",
        "title": title
    })

# ============================================================
# 2. LOAD MOTOR VEHICLES ACT
# ============================================================

!wget -q https://raw.githubusercontent.com/civictech-India/Indian-Law-Penal-Code-Json/main/MVA.json -O mva.json

with open("mva.json", "r", encoding="utf-8") as f:
    mva_data = json.load(f)

if isinstance(mva_data, list):
    for item in mva_data:
        section = str(item.get("section", item.get("Section", item.get("sec", "")))).strip()
        title = str(item.get("title", item.get("Title", ""))).strip()
        content = str(item.get("content", item.get("text", item.get("Content", str(item)))))[:600]

        text = f"Motor Vehicles Act Section {section} - {title}: {content}"
        documents.append(text)
        metadata.append({
            "act": "Motor Vehicles Act",
            "section": section if section else "N/A",
            "title": title
        })

elif isinstance(mva_data, dict):
    for key, value in mva_data.items():
        text = f"Motor Vehicles Act Section {key}: {value}"
        documents.append(text)
        metadata.append({
            "act": "Motor Vehicles Act",
            "section": str(key),
            "title": ""
        })

print(f"✅ Total documents loaded: {len(documents)}")

# ============================================================
# 3. EMBEDDINGS
# ============================================================

print("🔄 Creating embeddings...")

embedder = SentenceTransformer("all-MiniLM-L6-v2")

embeddings = embedder.encode(
    documents,
    show_progress_bar=True,
    convert_to_numpy=True
).astype("float32")

index = faiss.IndexFlatL2(embeddings.shape[1])
index.add(embeddings)

print(f"✅ FAISS index created with {index.ntotal} documents")

# ============================================================
# 4. MODEL
# ============================================================

print("🔄 Loading TinyLlama model...")

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True
)

model_name = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"

tokenizer = AutoTokenizer.from_pretrained(model_name)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    quantization_config=bnb_config,
    device_map="auto",
    torch_dtype=torch.float16
)

generator = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=180,
    do_sample=False,
    return_full_text=False,
    pad_token_id=tokenizer.eos_token_id
)

print("✅ Chatbot model loaded")

# ============================================================
# 5. HELPER FUNCTIONS
# ============================================================

def improve_query(query):
    q = query.lower()
    extra = []

    keyword_map = {
        "kill": "murder", "killed": "murder", "killing": "murder",
        "murder": "murder", "murdered": "murder", "death": "murder",
        "dead": "murder", "homicide": "culpable homicide", "died": "death murder",
        "muder": "murder", "panishment": "punishment", "punishment": "punishment",
        "injured": "hurt", "injury": "hurt", "hurt": "hurt",
        "blood": "hurt bleeding", "bleeding": "hurt", "wounded": "hurt",
        "wound": "hurt", "grievous": "grievous hurt", "attack": "hurt",
        "beaten": "hurt", "theft": "theft", "steal": "theft",
        "stole": "theft", "stolen": "theft", "robbery": "robbery",
        "rape": "rape", "raped": "rape", "sexual": "sexual offence",
        "sexual assault": "sexual assault", "helmet": "helmet protective headgear",
        "licence": "driving licence", "license": "driving licence",
        "drunk": "drunken driving", "alcohol": "drunken driving",
        "overspeed": "speed", "overspeeding": "speed",
        "seat belt": "seat belt", "seatbelt": "seat belt",
        "rash": "rash driving", "accident": "road accident",
        "hit and run": "hit and run", "kidnap": "kidnapping",
        "kidnapped": "kidnapping", "abduct": "abduction",
        "abducted": "abduction", "cheat": "cheating",
        "cheating": "cheating", "fraud": "cheating fraud",
        "dowry": "dowry", "dowry death": "dowry death",
        "cruelty": "cruelty", "harassment": "cruelty"
    }

    for k, v in keyword_map.items():
        if k in q:
            extra.append(v)

    return query + " " + " ".join(extra)


def detect_legal_intent(query):
    q = query.lower()
    intents = {
        "murder": ["kill", "killed", "killing", "murder", "murdered", "death", "dead", "died", "homicide", "muder"],
        "hurt": ["injured", "injury", "hurt", "blood", "bleeding", "wounded", "wound", "attack", "beaten", "grievous"],
        "rape": ["rape", "raped", "sexual assault", "sexual violence"],
        "theft": ["theft", "steal", "stolen", "stole", "robbery"],
        "traffic": ["helmet", "driving licence", "driving license", "license", "licence",
                    "overspeed", "overspeeding", "speeding", "drunk driving", "drunken driving",
                    "seat belt", "seatbelt", "traffic", "vehicle", "car accident", "bike accident",
                    "motor vehicle", "road accident"],
        "dowry": ["dowry", "dowry death"],
        "kidnapping": ["kidnap", "kidnapped", "kidnapping", "abduct", "abduction"],
        "fraud": ["fraud", "cheating", "cheated"]
    }

    detected = []
    for intent, keywords in intents.items():
        for keyword in keywords:
            if keyword in q:
                detected.append(intent)
                break
    return list(set(detected))


def legal_priority_filter(query, indices):
    intents = detect_legal_intent(query)

    if "murder" in intents:
        preferred = []
        for i in indices:
            act = metadata[i]["act"].lower()
            section = metadata[i]["section"].lower().strip()
            title = metadata[i]["title"].lower()
            if "bns" in act and (section in ["101", "103"] or "murder" in title or "culpable homicide" in title):
                preferred.append(i)
        if preferred:
            return preferred + [i for i in indices if i not in preferred]

    if "hurt" in intents:
        preferred = []
        for i in indices:
            act = metadata[i]["act"].lower()
            section = metadata[i]["section"][0:3].lower().strip()
            title = metadata[i]["title"].lower()
            if "bns" in act and (section in ["114", "115", "116", "117", "118"] or "hurt" in title):
                preferred.append(i)
        if preferred:
            return preferred + [i for i in indices if i not in preferred]

    if "traffic" in intents:
        preferred = []
        for i in indices:
            if "motor" in metadata[i]["act"].lower():
                preferred.append(i)
        if preferred:
            return preferred + [i for i in indices if i not in preferred]

    return list(indices)


def remove_irrelevant_context(query, indices):
    intents = detect_legal_intent(query)
    filtered = []

    for i in indices:
        act = metadata[i]["act"].lower()
        title = metadata[i]["title"].lower()

        # Strongly reject maintenance / family sections when asking about murder
        if "murder" in intents:
            if "maintenance" in title or "145" in metadata[i]["section"] or "family" in title:
                continue
            if "motor vehicles act" in act and "traffic" not in intents:
                continue

        if "traffic" in intents and "murder" not in intents and "hurt" not in intents:
            if "motor" not in act and "vehicle" not in act:
                continue

        filtered.append(i)

    return filtered if filtered else list(indices)


def find_priority_sections(query):
    intents = detect_legal_intent(query)
    priority = []

    for i, item in enumerate(metadata):
        act = item["act"].lower()
        section = item["section"].lower().strip()
        title = item["title"].lower()

        if "murder" in intents and "bns" in act:
            if section in ["101", "103"] or "murder" in title or "culpable homicide" in title:
                priority.append(i)

        if "hurt" in intents and "bns" in act:
            if section in ["114", "115", "116", "117", "118"] or "hurt" in title:
                priority.append(i)

    return list(dict.fromkeys(priority))


def is_greeting(query):
    q = query.lower().strip()
    greetings = [
        "hi", "hello", "hey", "hii", "hiii", "helo", "hola",
        "good morning", "good afternoon", "good evening", "good night",
        "namaste", "namaskar", "how are you", "how r u", "what's up", "whats up",
        "thank you", "thanks", "thank u", "thx", "bye", "goodbye"
    ]
    for g in greetings:
        if q == g or q.startswith(g + " ") or q.startswith(g + "!") or q.startswith(g + "?"):
            return True
    return False


def get_greeting_reply(query):
    q = query.lower().strip()
    if any(w in q for w in ["hi", "hello", "hey", "hii", "namaste"]):
        return ("Hello! 👋 Welcome to the *Indian Legal Assistant*.\n\n"
                "I can help you with questions related to *BNS, BNSS, BSA* and the *Motor Vehicles Act*.\n\n"
                "How can I assist you today?")
    if "how are you" in q:
        return "I'm doing great, thank you! Ready to help with your legal questions."
    if any(w in q for w in ["thank", "thanks"]):
        return "You're welcome! Feel free to ask anything related to Indian law."
    if any(w in q for w in ["bye", "goodbye"]):
        return "Goodbye! Take care. You can come back anytime."
    return "Hello! How can I help you with Indian law today?"


# ============================================================
# 6. MAIN ASK FUNCTION (Improved Answer Format)
# ============================================================

def ask(query, history=[]):
    query = query.strip()
    if not query:
        return "Please type a question."

    # ---------- Greeting ----------
    if is_greeting(query):
        return get_greeting_reply(query)

    # ---------- Legal Pipeline ----------
    search_text = improve_query(query)
    intents = detect_legal_intent(query)

    if "murder" in intents:
        search_text += " BNS Section 101 Section 103 murder culpable homicide punishment"
    if "hurt" in intents:
        search_text += " BNS hurt Section 115 grievous hurt Section 116"
    if "traffic" in intents:
        search_text += " Motor Vehicles Act traffic driving licence helmet"
    if "rape" in intents:
        search_text += " BNS rape sexual offence"
    if "theft" in intents:
        search_text += " BNS theft stolen property"

    q_emb = embedder.encode([search_text], convert_to_numpy=True).astype("float32")
    distances, indices = index.search(q_emb, 20)

    filtered_indices = legal_priority_filter(query, indices[0])
    filtered_indices = remove_irrelevant_context(query, filtered_indices)

    priority_sections = find_priority_sections(query)
    final_indices = list(dict.fromkeys(priority_sections + filtered_indices))[:6]

    if not final_indices:
        final_indices = list(indices[0][:4])

    context_parts = []
    sources = []
    for i in final_indices:
        context_parts.append(documents[i])
        sources.append(metadata[i])

    context = "\n\n".join(context_parts)

    prompt = f"""You are an Indian Legal Assistant specialized in BNS 2023.

Answer the question using ONLY the legal context given below.

Rules:
- Be direct and clear.
- Always mention Act name + Section number.
- For murder questions, focus on BNS Section 101 and Section 103.
- Use simple English.
- Keep answer between 4 to 7 sentences.
- Do not invent sections.

Legal Context:
{context}

Question: {query}

Answer:"""

    try:
        result = generator(prompt)
        answer = result[0]["generated_text"].strip()

        if "Answer:" in answer:
            answer = answer.split("Answer:")[-1].strip()

        answer = re.sub(r"^(Answer:|Response:)\s*", "", answer, flags=re.I).strip()
        answer = re.sub(r"\n{3,}", "\n\n", answer).strip()

        if not answer or len(answer) < 20:
            answer = "Based on the available legal provisions, more specific facts are needed for a precise answer."

    except Exception as e:
        answer = "Sorry, I could not generate a proper legal answer right now. Please try again."

    # ---------- Build Final Clean Response ----------
    final_response = f"*Answer:*\n{answer}\n\n"

    # Detected Intent
    if intents:
        final_response += f"*Detected Legal Intent:* {', '.join(intents).title()}\n\n"
    else:
        final_response += "*Detected Legal Intent:* General Legal Query\n\n"

    # Relevant Sections
    shown = set()
    source_lines = []
    for s in sources:
        key = (s["act"], s["section"])
        if key not in shown and s["section"] != "N/A":
            source_lines.append(f"• {s['act']} → Section {s['section']}")
            shown.add(key)
        if len(shown) >= 5:
            break

    if source_lines:
        final_response += "*Relevant Sections:*\n" + "\n".join(source_lines) + "\n\n"
    else:
        final_response += "*Relevant Sections:* Not clearly identified from context.\n\n"

    final_response += "⚠️ *Disclaimer:* This is general legal information only and is not a substitute for advice from a qualified lawyer."

    return final_response


# ============================================================
# 7. LAUNCH
# ============================================================

print("\n🚀 Launching Chat Window...")

demo = gr.ChatInterface(
    fn=ask,
    title="⚖️ Indian Legal Assistant",
    description="*BNS • BNSS • BSA • Motor Vehicles Act*\n\nAsk any legal question. You can also say Hi / Hello.",
    examples=[
        "What is the punishment for murder under BNS?",
        "murder case punishment",
        "What is grievous hurt under BNS?",
        "Helmet rules under Motor Vehicles Act",
        "Punishment for theft under BNS",
        "What is the fine for drunk driving?",
        "Punishment for rape under BNS",
        "hit and run case",
        "Dowry death provisions",
        "Hi"
    ],
    chatbot=gr.Chatbot(height=520),
    textbox=gr.Textbox(
        placeholder="Type your question here... (e.g. murder case punishment)",
        container=False,
        scale=7
    ),
)

demo.launch(
    share=True,
    theme=gr.themes.Soft(primary_hue="indigo", secondary_hue="blue")
)
```
